### Model Evaluation

Write a function to generate the confusion matrix for a list of actual classes and a list of predicted classes

```python
from setuptools.package_index import unique_values

actual_class = ["Yes", "No", "No", "Yes", "No", "No", "Yes", "No", "No", "No"]
predicted_class = ["Yes", "No", "Yes", "No", "No", "No", "Yes", "Yes", "Yes", "No"]

def confusion_matrix(actual, predicted):
    if (len(actual) != len(predicted)):
        raise ValueError

    unique_classes=sorted(list(set(actual+predicted)))
    matrix = { (act, pred): 0 for act in unique_classes for pred in unique_classes }

    for act, pred in zip(actual, predicted):
        matrix[(act, pred)] += 1

    return matrix

print(confusion_matrix(actual_class, predicted_class))
```
